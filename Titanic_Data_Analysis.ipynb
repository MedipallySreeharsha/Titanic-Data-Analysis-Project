# Titanic Data Analysis
Exploratory analysis of survival patterns using Pandas, Matplotlib and Seaborn.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('titanic.csv')
df.head()
```

```python
print('Shape:', df.shape)
display(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
print('Overall survival rate:', round(df['Survived'].mean()*100, 2), '%')
```

```python
print('Survival by sex:')
display(df.groupby('Sex')['Survived'].mean().mul(100).round(2))
print('Survival by class:')
display(df.groupby('Pclass')['Survived'].mean().mul(100).round(2))
print('Survival by port:')
display(df.groupby('Embarked')['Survived'].mean().mul(100).round(2))
```

```python
sns.countplot(data=df, x='Survived')
plt.title('Titanic Survival Count')
plt.show()
```

```python
sns.barplot(data=df, x='Sex', y='Survived')
plt.title('Survival Rate by Sex')
plt.show()
```

```python
sns.barplot(data=df, x='Pclass', y='Survived')
plt.title('Survival Rate by Passenger Class')
plt.show()
```

```python
sns.histplot(data=df, x='Age', bins=30, kde=True)
plt.title('Age Distribution')
plt.show()
```

```python
sns.boxplot(data=df, x='Survived', y='Fare')
plt.title('Fare Distribution by Survival')
plt.show()
```
